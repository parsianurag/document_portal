# Data Ingestion

```python
from langchain_community.document_loaders import PyPDFLoader
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
import os
```

```python
file_path=os.path.join(os.getcwd(),"data","sample_data.pdf")# here is my data path pdf
                #getcwd() get_current_working_directory
```

```python
os.getcwd()# see you get current working directory
```

```python

#Read the data from pdf
loader=PyPDFLoader(file_path)
```

```python
documents=loader.load()
```

```python
documents
```

```python
len(documents)# every document is conisdered as a one page
```

# Chuncking


```python
#this is an experimental there is no deterministic way to split the text
text_splitter=RecursiveCharacterTextSplitter(
    chunk_size=100,
    chunk_overlap=50,
    length_function=len
)
```

```python
docs=text_splitter.split_documents(documents)
```

```python
len(docs) # now it as more documents after splitting
          #note it is only for text data (not for images, voice, video data)
```

```python
docs[0].metadata # neta data
```

```python
docs[2].page_content
```

# Embedding

```python
from langchain_groq import ChatGroq
```

```python
from dotenv import load_dotenv
load_dotenv()
```

```python
model=ChatGroq(model="qwen/qwen3-32b")
```

```python
from langchain_google_genai import GoogleGenerativeAIEmbeddings
```

```python
embedding_model=GoogleGenerativeAIEmbeddings(model="text-embedding-004")
```

```python
embedding_model.embed_query("Hello world")
```

```python
embedding_model.embed_documents(docs[0].page_content)
```

```python
len(embedding_model.embed_documents(docs[0].page_content)[0])# length of embedding
```

```python
from langchain_community.vectorstores import FAISS
```

```python
vectorstore=FAISS.from_documents(docs,embedding_model) # it will store the data in vectorization form
    # this (faiss) in inmemory database stores in RAM
    # but this (faiss) you can persist it to disk as well
    # but this (faiss) us not on cloud database
```

## this is a retrival process
means from the vectordatabase we are going to fetch or retrive or rank the most appropriate k result

```python
vectorstore.similarity_search("What is a llama2 llm model")
```

```python
relevant_doc=vectorstore.similarity_search("What is a llama2 llm model")
```
